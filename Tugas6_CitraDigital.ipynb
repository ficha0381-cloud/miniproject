```python
import cv2
import numpy as np
import glob
import os
from google.colab.patches import cv2_imshow

# 1. Fungsi Deteksi Tanda Tangan
def detect_signature(image_path):

    img = cv2.imread(image_path)

    if img is None:
        print(f"Gagal membaca {image_path}")
        return

    # Crop area kanan atas
    h, w = img.shape[:2]

    y_start = int(h * 0.10)
    y_end   = int(h * 0.35)
    x_start = int(w * 0.65)
    x_end   = int(w * 0.95)

    cropped_img = img[y_start:y_end, x_start:x_end]

    # Grayscale
    gray = cv2.cvtColor(cropped_img, cv2.COLOR_BGR2GRAY)

    # Global Threshold
    _, global_thresh = cv2.threshold(
        gray, 127, 255, cv2.THRESH_BINARY_INV
    )

    # Otsu Threshold
    _, otsu_thresh = cv2.threshold(
        gray, 0, 255,
        cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU
    )

    # Morfologi
    kernel = np.ones((3, 3), np.uint8)

    opening = cv2.morphologyEx(
        otsu_thresh,
        cv2.MORPH_OPEN,
        kernel
    )

    closing = cv2.morphologyEx(
        opening,
        cv2.MORPH_CLOSE,
        kernel
    )

    # Perhitungan piksel
    foreground_pixels = cv2.countNonZero(closing)

    total_pixels = closing.shape[0] * closing.shape[1]

    pixel_ratio = (foreground_pixels / total_pixels) * 100

    # Keputusan
    if pixel_ratio > 1.5:
        status = "SIGNATURE PRESENT"
    else:
        status = "SIGNATURE ABSENT"

    # Hasil
    print("=" * 50)
    print(f"File              : {os.path.basename(image_path)}")
    print(f"Ukuran Asli       : {w} x {h} piksel")
    print(f"Jumlah Piksel TTD : {foreground_pixels}")
    print(f"Rasio Piksel      : {pixel_ratio:.2f}%")
    print(f"Status            : {status}")

    print("\nHasil Crop Area Kanan Atas:")
    cv2_imshow(cropped_img)

    print("\n")


# ==========================================================
# 2. MEMBACA GAMBAR DI FOLDER /content
# ==========================================================

test_images = (
    glob.glob("/content/*.jpg") +
    glob.glob("/content/*.jpeg") +
    glob.glob("/content/*.png")
)

if len(test_images) == 0:

    print("Tidak ada gambar ditemukan di /content!")

else:

    print(f"Memproses {len(test_images)} gambar...\n")

    for img_path in sorted(test_images):
        detect_signature(img_path)
```
